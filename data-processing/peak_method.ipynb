# Get Conductivities by peak estimation
These are imprecise conductivities but by using the peak method we can see the trend more clearly. for troubleshooting
## First obtain average transmission ratio and standard deviations

```python
import matplotlib.pyplot as plt
import numpy as np
import numpy.fft as fft
from pprint import pprint
import json
import os
%matplotlib widget

#file_root = '/Users/yuewang/SLAC/goldthz/Au_data/800nm/processed_data/single_shot/' # 800 nm
file_root = '/Users/yuewang/SLAC/goldthz/Au_data/400nm/processed_data/single_shot/' # 400 nm

wp = 'wp43'
plt.close('all')

allRatios = {}
interRatios = []
def getRatio(path, wp, pos, num):
    global interRatios
    if num == '0001':
        interRatios = []

    file_path = file_root + path
    post_bal_raw = np.loadtxt(file_path + 'THz_post_bal.dat')
    post_bal = np.mean(post_bal_raw, axis=1)  # average over the columns
    sig_bal_raw = np.loadtxt(file_path + 'THz_pre_bal.dat')
    sig_bal = np.mean(sig_bal_raw, axis=1)
    time_axis = np.loadtxt(file_path + 'time_axis.dat')

    plt.figure(figsize=(10, 6))
    plt.plot(time_axis, post_bal, "b-",label="THz post") # plot only the 1st column of post_bal
    plt.plot(time_axis, sig_bal, "r-",label="THz pre")
    plt.xlabel("Time (ps)")
    plt.ylabel("THz signal (unit)")
    plt.title(f"THz signal comparison for {wp}, {pos}, shot {num}")
    plt.legend()

    pMask = (time_axis >= 1.638) & (time_axis <= 1.6900)
    sMask = (time_axis >= 1.5714) & (time_axis <= 1.6837)
    pMaxMean = np.mean(post_bal[pMask])
    sMaxMean = np.mean(sig_bal[sMask])
    plt.axhline(y = pMaxMean)
    plt.axhline(y=sMaxMean)
    plt.show()

    ratio = sMaxMean/pMaxMean
    interRatios.append(ratio)

    if num == '0003' or num == '0007':
        print(interRatios)
        meanRatios = np.mean(interRatios)
        stdRatios = np.std(interRatios)
        allRatios[pos] = {"mean": meanRatios, "std": stdRatios}

    print("ratio: ", ratio)


def cycle_through(position: str, p_range, wp='wp43'):
    for digit in p_range:
        pos = position + str(digit)

        shot_num = 7
        for shot in range(shot_num):
            num = '{:0>4}'.format(int(shot + 1))
            #path = 'SS_Au800_' + wp + '_' + pos + '_shot_' + num + "/" # 800 nm
            path = 'SS_Au_' + wp + '_' + pos + '_shot_' + num + "/" # 400 nm
            file_path = file_root + path

            print("--------------------------------")
            print(f"File path: {file_path}")
            if not os.path.exists(file_path):
                print(f"File path does not exist: {file_path}")
                continue
            else:
                print(f"creating plot for file: {file_path}")
                getRatio(path, wp, pos, num)
cycle_through("neg", [i for i in range(3)], wp='wp41') # 400 nm
cycle_through("pos", [1, 2, 5, '7p5', 10, '12p5', 15, 25, 50, 100], wp='wp41') # for 400 nm 

#cycle_through("neg", [i for i in range(3)], wp='wp43') # 800 nm
#cycle_through("pos", [1, 2, 5, '7p5', 10, 12, 15, 20, 25, 30, 50]) # for 800 nm

pprint(allRatios)
```

## Estimate Conductivity

```python
# use the Ratio data to estimate conductivity
# sigma vars

SAVE = True
Z0 = 376.73
nSi3N4 = 2.7

sigmaData = {} # for storing mean sigmas at each pos

def saveDict(data, file_path):
    with open(file_path, "w") as file:
        json.dump(data, file, indent=4)

def sigmaCycle(position: str, p_range):
    for digit in p_range:
        pos = position + str(digit)
        thickness = 30e-9
        transfer = allRatios[pos]["mean"]
        stdev = allRatios[pos]["std"]
        A = ((1 + nSi3N4) / (Z0*thickness))
        sigma = A * ((1/transfer) - 1)
        stdSigma = A * stdev / (transfer**2)
        print(sigma, stdSigma)
        sigmaData[pos] = {"mean": sigma, "std": stdSigma}

sigmaCycle("neg", [i for i in range(3)])
#sigmaCycle("pos", [1, 2, 5, '7p5', 10, 12, 15, 20, 25, 30, 50]) # 800 nm
sigmaCycle("pos", [2, 5, '7p5', 10, '12p5', 15, 25, 50, 100]) # 400 nm

def renameKeys(old_data): # rename keys to float, then sort by key
    new_data = {}
    for key, value in old_data.items():
        if key.startswith('neg'):
            key = key.replace('neg', '-')
            new_key = float(key)
        elif key.startswith('pos'):
            key = key.replace('pos', '')
            if 'p' in key:
                key = key.replace('p', '.')

            new_key = float(key)
        else:
            new_key = key
        new_data[new_key] = value
    sort_data = dict(sorted(new_data.items()))
    return sort_data

sortedSigmaData = renameKeys(sigmaData)
pprint(sortedSigmaData)
if SAVE:
    saveDict(sortedSigmaData, '/Users/yuewang/SLAC/goldthz/Au_data/code/' + 'Peak method/' + '400_nm_peaks_pre.json')


# now plot all the conductivities!

plt.figure(figsize=(9,6))
yErr_peak = [v["std"] for v in sortedSigmaData.values()]
y_peak = [v["mean"] for v in sortedSigmaData.values()]
time800 = list(sortedSigmaData.keys())
plt.errorbar(list(sortedSigmaData.keys()), y_peak, fmt="-ob", yerr=yErr_peak, capsize=5, capthick=2, ms=7, label=r"$Peak method: \lambda=800nm, U=0.26J/m^3$", lw=3)
plt.xlabel("Pump-probe delay (ps)", fontsize=18, weight='bold')
plt.ylabel(r"$\mathbf{\sigma_{THz}(\omega)\:(S/m)}$", fontsize=18)
plt.xlim(-3, 50)
plt.yticks(fontsize=16)
plt.tick_params(which="both", axis="both", labelsize=14)
plt.title("Conductivity vs delay for Au")

plt.xticks(fontsize=16)
plt.yscale('log')
plt.legend()
```

## Compare to FFT obtained conductivity
This FFT has the exponential decay after 5 ps

```python
positions = ['neg0', 'neg1', 'neg2', 'pos1', 'pos2', 'pos5', 'pos7p5', 'pos10', 'pos12', 'pos15', 'pos20', 'pos25', 'pos30', 'pos50']

for pos in positions:
    realSigmas = [] # for calculating means across shots for each pos

    for shot in range(7):
        if not os.path.exists(file_root + f'SS_Au800_{wp}_{pos}_shot_{shot:04d}/'):
            continue

        file_path = file_root + f'SS_Au800_{wp}_{pos}_shot_{shot:04d}/'
        try:
            # load data -- pre shot
            post_bal = np.loadtxt(file_path + 'dec_post.dat')
            shot_bal = np.loadtxt(file_path + 'dec_shot.dat')
            time_axis = np.loadtxt(file_path + 'dec_time.dat')
        except FileNotFoundError as e:
            print(f"File not found: {e}")
            continue

        # compute frequency axis 
        dt = time_axis[1] - time_axis[0]
        N = len(time_axis)

        # zero padding
        Nfft = 5336 # 8 times number of data points

        # padded fft
        ref_fft = fft.fft(post_bal, n=Nfft)
        sig_fft = fft.fft(shot_bal, n=Nfft)
        trans_fft = sig_fft / ref_fft

        freq = fft.fftfreq(Nfft, dt)
        # sort by frequency
        index = np.argsort(freq)
        frq_sort = freq[index]
        ref_sort = ref_fft[index]
        sig_sort = sig_fft[index]

        # finding the mean sigma in a freq range per shot
        sThickness = 30e-9
        transfer = sig_sort/ref_sort
        sigma = ((1 + nSi3N4) / (Z0*sThickness)) * ((1/transfer) - 1)
        sigmaReal = np.real(sigma)
        sigmaSection = np.mean(sigma[(frq_sort >= 0.73) & (frq_sort <= 0.9)])
        realSigmas.append(np.real(sigmaSection))

    print(f"sigma values at {pos}: {realSigmas}")
        
    realSigmaMean = np.mean(realSigmas) # realSigmas holds the mean sigma for each shot at a delay
    realSigmaStd = np.std(realSigmas)
    sigmaData[pos] = {"mean": realSigmaMean, "std": realSigmaStd}
sortedSigmaData = renameKeys(sigmaData)
print("sorted procesed conductivities")
pprint(sortedSigmaData)
# conductivity vs delay figure 
plt.figure(figsize=(9,6))
yErr_5ps = [v["std"] for v in sortedSigmaData.values()]
y_5ps = [v["mean"] for v in sortedSigmaData.values()]
```
